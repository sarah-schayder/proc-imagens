```python
import skimage as ski
import os
from natsort import natsorted, ns
```

```python
caminho = "/content/drive/MyDrive/processamento de imagens/"
```

```python
list_files = os.listdir("/content/drive/MyDrive/processamento de imagens")
list_files
```

```python
list_files = natsorted(list_files)
list_files
```

```python
image_list = []
for filename in list_files:
  image_list.append(ski.io.imread(caminho + filename))
```

```python
for i in image_list:
  ski.io.imshow(i)
  ski.io.show()
```
