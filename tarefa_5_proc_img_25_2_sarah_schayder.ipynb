```python
import skimage as ski
import os
from natsort import natsorted, ns
import numpy as np
import matplotlib.pyplot as plt
```

```python
list_files = os.listdir("/content/drive/MyDrive/processamento de imagens/nivel-de-cinza/")
list_files
```

```python
caminho = "/content/drive/MyDrive/processamento de imagens/nivel-de-cinza/"
```

```python
image_list = []
for filename in list_files:
  image_list.append(ski.io.imread(caminho + filename))
```

```python
caminho2 = "/content/drive/MyDrive/processamento de imagens/imagens-tarefa5"
```

```python
for i, img_name in enumerate(image_list):
  filename = list_files[i]

  # Aplicando a equalização de histograma
  eq_hist = ski.exposure.equalize_hist(img_name)

  # Mostrando as imagens
  fig, axs = plt.subplots(1, 2, figsize=(15, 5))
  axs[0].imshow(img_name, cmap='gray')
  axs[0].set_title('Nível de cinza')
  axs[1].imshow(eq_hist, cmap='gray')
  axs[1].set_title('Equalização de Histograma')
  plt.show()

  # Salvando as novas imagens
  eqhst_filename = os.path.join(caminho2, 'eq-hist_' + filename)

  ski.io.imsave(eqhst_filename, ski.util.img_as_ubyte(eq_hist))
```
