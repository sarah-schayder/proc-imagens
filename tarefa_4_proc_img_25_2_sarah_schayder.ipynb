```python
import skimage as ski
import os
from natsort import natsorted, ns
import numpy as np
import matplotlib.pyplot as plt
```

```python
caminho = "/content/drive/MyDrive/processamento de imagens/nivel-de-cinza/"
```

```python
list_files = os.listdir("/content/drive/MyDrive/processamento de imagens/nivel-de-cinza/")
list_files
```

```python
image_list = []
for filename in list_files:
  image_list.append(ski.io.imread(caminho + filename))
```

```python
caminho2 = "/content/drive/MyDrive/processamento de imagens/imagens-tarefa-4"
```

```python
for i, img_name in enumerate(image_list):
  filename = list_files[i]

  # Tranformando em negativo
  negativo = ski.util.invert(img_name)

  # Transformando em Gamma
  gamma = ski.exposure.adjust_gamma(img_name, gamma=0.5)  # gamma < 1 clareia, >1 escurece

  # Mostrando as imagens
  fig, axs = plt.subplots(1, 3, figsize=(15, 5))
  axs[0].imshow(img_name, cmap='gray')
  axs[0].set_title('Nível de cinza')
  axs[1].imshow(negativo, cmap='gray')
  axs[1].set_title('Negativo')
  axs[2].imshow(gamma, cmap='gray')
  axs[2].set_title('Gamma')
  plt.show()

  # Salvando as novas imagens
  negativo_filename = os.path.join(caminho2, 'negativo_' + filename)
  gamma_filename = os.path.join(caminho2, 'gamma_' + filename)

  ski.io.imsave(negativo_filename, ski.util.img_as_ubyte(negativo))
  ski.io.imsave(gamma_filename, ski.util.img_as_ubyte(gamma))
```
