```python
import skimage as ski
import os
from natsort import natsorted, ns
import numpy as np
import matplotlib.pyplot as plt
```

```python
caminho = "/content/drive/MyDrive/processamento de imagens/"
```

```python
img = ski.io.imread(caminho + "flor_o_d_sarah.jpg")
plt.imshow(img)
```

```python
img2 = ski.color.rgb2gray(img)

plt.imshow(img2)
```

```python
img2.shape
```

```python
img2[0,0]
```

```python
neg = 255 - img2
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img2)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(neg)
ax[1].set_title("Negativo")
ax[1].axis("off")

plt.show()
```
