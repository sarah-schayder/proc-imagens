```python
import skimage as ski
import os
from natsort import natsorted, ns
import matplotlib.pyplot as plt
```

```python
caminho = "/content/drive/MyDrive/processamento de imagens/banco-imgs-original/"
```

```python
list_files = os.listdir("/content/drive/MyDrive/processamento de imagens/banco-imgs-original/")
list_files
```

```python
image_list = []
for filename in list_files:
  image_list.append(ski.io.imread(caminho + filename))
```

```python
caminho2 = "/content/drive/MyDrive/processamento de imagens/nivel-de-cinza/"
```

```python
for filename, image in zip(list_files, image_list):
  gray_image = ski.color.rgb2gray(image)
  ski.io.imsave(caminho2 + 'cinza_' + filename, ski.util.img_as_ubyte(gray_image))
  plt.imshow(gray_image, cmap='gray')
  plt.show()
```
